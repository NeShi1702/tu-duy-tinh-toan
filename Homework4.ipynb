W4A18 Nhập vào số nguyên dương a và b, in toàn bộ ước chung của a và b 


```python
a = int(input())
b = int(input())
for i in range(1, 10**10):
    if a % i == 0 and b % i == 0:
        print(i)
    elif i > a or i > b:
        break
```

W4A21 Nhập vào số nguyên dương n, tính tổng các chữ số của n 


```python
n = int(input())
tong = 0
while n > 0:
    tong += n % 10
    n //= 10
print(tong)
```

W4A22 Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu chữ số lẻ 


```python
n = int(input())
so_chan = 0
so_le = 0
if n == 0:
    print("1 0")
while n > 0:
    chu_so = n % 10
    if chu_so % 2 == 0:
        so_chan += 1
    else:
        so_le += 1
    n //= 10
print(so_chan, so_le)
```

W4A23 
Nhập n
Cho S(k) = 1 + 2 + 3 + … + k
Tìm k sao cho S(k) lớn nhất nhưng nhỏ hơn n 


```python
n = int(input())
S = 0
k = 0
while S < n:
    k += 1
    S += k
print(k - 1)
```

W4A24
Nhập vào A, tìm n nhỏ nhất sao cho 1 + 1/2 + 1/3 + 1/4 + ... + 1/n > A

```python
n = int(input())
S = 0
k = 0
while S < n:
    k += 1
    S += 1 / k
print(k - 1) 
```
