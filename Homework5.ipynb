Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số nguyên x
<Số Amstrong là số = tổng các chữ số và mỗi số đc lũy thừa = tổng số các chữ số (VD: 1634 = 1^4 +6^4 + 3^4 + 4^4)>

```python
def Amstrong(n) -> bool:
    s = 0
    for i in str(n):
        s += int(i) ** len(str(n))
    return s == n
a = int(input())
print(Amstrong(a))
```

W5A20 Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'.

W5A23 Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình a*x^2 + b*x + c = 0.

```python
def quadatic_roots_count(a, b, c):
    delta = b**2 - 4*a*c
    if delta > 0:
        return 2
    elif delta == 0:
        return 1
    else:
        return 0
a = int(input())
b = int(input())
c = int(input())
print(quadatic_roots_count(a, b, c))
```

W5A24 Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không.


```python
def are_collonear(x1, y1, x2, y2, x3, y3):
    if (y2 - y1) / (y3 - y2) == (x2 - x1) / (x3 - x2):
        return("Thẳng hàng")
    else:
        return("Không thẳng hàng")
a = int(input())
b = int(input())
c = int(input())
d = int(input())
e = int(input())
f = int(input())
print(are_collonear(a, b, c, d, e, f))
```

W5A25 Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay
không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3),
(0,3)], p = (2,1)

```python
def point_in_quadrilateral(x1, y1, x2, y2, x3, y3, x4, y4, x, y):
    area1 = 0.5 * abs(x1*(y2-y3) + x2*(y3-y1) + x3*(y1-y2))
    area2 = 0.5 * abs(x1*(y2-y4) + x2*(y4-y1) + x4*(y1-y2))
    area3 = 0.5 * abs(x1*(y3-y4) + x3*(y4-y1) + x4*(y1-y3))
    area4 = 0.5 * abs(x2*(y3-y4) + x3*(y4-y2) + x4*(y2-y3))
    total_area = area1 + area2 + area3 + area4
    quadrilateral_area = 0.5 * abs(x1*(y2-y3) + x2*(y3-y4) + x3*(y4-y1) + x4*(y1-y2))
    return total_area == quadrilateral_area
a = int(input())
b = int(input())
c = int(input())
d = int(input())
e = int(input())
f = int(input())
g = int(input())
h = int(input())
x = int(input())
y = int(input())
print(point_in_quadrilateral(a, b, c, d, e, f, g, h, x, y))
```

W5A28 Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3).


```python
def goc_ABC(x1, y1, x2, y2, x3, y3):
    AB = ((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5
    AC = ((x3 - x1) ** 2 + (y3 - y1) ** 2) ** 0.5
    ABxAC = (x2 - x1) * (x3 - x1) + (y2 - y1) * (y3 - y1)
    cosABC = ABxAC / (AB * AC)
    return cosABC
x1, y1, x2, y2, x3, y3 = map(int, input().split())
print(goc_ABC(x1, y1, x2, y2, x3, y3))
```

W5A29 Viết hàm trả về giá trị lớn nhất và nhỏ nhất của ba số a,b,c dưới dạng list [max, min]

W5A30 Viết hàm tính hệ số nhị thức C(n, k).


```python
def he_so_nhi_thuc_C(n, k):
    if k > n:
        return 0
    elif k == 0 or k == n:
        return 1
    else:
        C = 1
        for i in range(1, k + 1):
            C = C * (n - i + 1) // i
        return C
n, k = map(int, input().split())
print(he_so_nhi_thuc_C(n, k))
```

W5A33 Viết hàm smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính tổng tiền taxi theo quy tắc giá mở cửa,
giá km, phí chờ, phụ phí ban đêm và giảm giá.

W5A34 Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau
Nếu n ≤ 0 trả về 0.

```python
def sum_1_to_n(n):
    tong = 0
    for i in range(1, n + 1):
        tong += i 
    return tong
n = int(input())
print(sum_1_to_n(n))
```
